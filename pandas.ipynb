## Filtering and boolean indexing

```python
df
```

```python
df[df['age']>28] #displays the values which are greater than 28
```

```python
df
```

```python
df[(df['age']>25)&(df['city']=='NY')] #multiple conditions combine with & and |
```

```python
df[(df['age']>25)&(df['age']<35)]
```

```python
df[(df['age']>35)|(df['city']=='NY')]
```

```python
df[(df['name']=='Bob')|(df['city']=='NY')]
```

```python
#isin -->filters if the value is present in the table or not
df[df['city'].isin(['NY','LA'])]
```

```python
#.query-->filters using readable string expression
df.query('age>28 and city=="NY"')
```

```python
#.where-->keep matching values replace other with NaN
df.where(df['age']>28)
```

```python
df.mask(df['age']>28) 
```

## Handling missing values

```python
df_na=pd.DataFrame({'a':[1,np.nan,3],'b':[np.nan,2,3]})
df_na
```

```python
#  .isna() ---> Boolean mask of missing values
df_na.isna()
```

```python
# .notna() ----> opposite of .isna()
df_na.notna()
```

```python
#df_na.isnull().sum() ---> Count of missing values per columns
df_na.isnull().sum()
```

```python
# .dropna()-----> Drop rows containing any NaN
df_na.dropna()
```

```python
# .dropna(axis=1)-----> Drop columns containing any NaN
df_na.dropna(axis=1)
```

```python
#  .fillna() ---->  fill missing values with constant
df_na.fillna(0)
```

```python
#   .ffill() -----> Forward fill missing values
df_na.ffill()
```

```python
#   .bfill() -----> Backward fill missing values
df_na.bfill()
```

## Sorting And Ranking

```python
df=pd.DataFrame({
    'name':['Alice','Bob','Charlie','David'],
    'age':[25,30,35,40],
    'city':['NY','LA','SF','NY']
})
df
```

```python
# .sort_values() ----> Arrange the values in ascending order
df.sort_values('age')
```

```python
df.sort_values(['city','age'],ascending=[True,False])
```

```python
# .sort_index() ----> Sort rows by the index
df.sort_index()
```

```python
# .rank() ----> Rank values within a column
df['age'].rank
```

```python
# .nlargest() ----> Top n rows by a column
df.nlargest(2,'age')
```

```python
# .nsmallest()----> Bottom n rows by a column
df.nsmallest(3,'age')
```

## Aggregation and Statistics

```python
# .sum() / .mean() / .median() ------> Basic Aggregations
df['age'].sum(), df['age'].mean(), df['age'].median()
```

```python
# .std() / .var() ----> Spread of data
df['age'].std(), df['age'].var()
```

```python
# .min() / .max() ----> Smallest and Largest value
df['age'].min(), df['age'].max()
```

```python
# .count() ---> Number of not null values
df['age'].count()
```

```python
# .agg()----> Apply a multiple aggregation functions at once
df.agg({'age':['mean','max','min']})
```

```python
# .corr() ----->correlation matrix of numeric columns
df.corr(numeric_only=True)
```

```python
# .mode() ---->Most frequently occuring value
df['city'].mode()
```

## GroupBy Operations

```python
g=df.groupby('city')
```

```python
#  .size()---> Number of rows in each group
g.size()
```

```python
# .agg()---> On a group by -- Multiple aggregations per group
```

```python
g.agg({'age':['mean','sum']})
```

```python
#  .apply()---> on a groupby -- custom function per group
g.apply(lambda x: x['age'].max() - x['age'].min())
```

```python
add=lambda a,b:a+b
```

```python
add(2,2)
```

```python
Iterating over groups with a for loop
```

```python
for name, group in g:
    print(name)
    print(group)
```

```python
# .get_group() ---> Retrieve a specific group
g.get_group('NY')
```
