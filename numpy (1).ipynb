### Random Module (np.random) --->This model helps us in replicating the data

```python
import numpy as np
```

```python
#This will generate any random numbers
np.random.rand(10)
```

```python
#np.random.randn()--> This will generate the values that lie between -3 to +3   i.e.normal distribution 
np.random.randn(10)
```

```python
#np.random.randint()--->This will generate integer values
np.random.randint(1,20,5)       #1- starting value , 20 is ending value and 5 is size 
```

```python
#np.random.uniform()--->uniform distribution with given mean / SD
np.random.uniform(0,5,10)
```

```python
#np.random.normal()-----> Normal distribution with given mean / SD
np.random.normal(0,1,5)
```

```python
#np.random.choice()----->Random sample from a given array
np.random.choice([1,2,3,4],size=3, replace=False)    # false is used so that the number should not be repeated
```

```python
#np.random.shuffle()---->Shuffle an array in place
arr=np.arange(5)
np.random.shuffle(arr)
arr
```

```python
#np.random.permutation()----> Return a shuffled copy
np.random.permutation(10)
```
